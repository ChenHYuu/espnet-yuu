```python
import os 
import whisper
import numpy as np
import torch
import opencc
from tqdm.notebook import tqdm
from time import sleep
```

```python
# DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DEVICE = "cpu"
```

```python
model = whisper.load_model("base", device=DEVICE)

# Simple to Traditional Chinese
converter = opencc.OpenCC('s2t.json')
```

```python
print(
    f"Model is {'multilingual' if model.is_multilingual else 'English-only'} "
    f"and has {sum(np.prod(p.shape) for p in model.parameters()):,} parameters."
)
```

```python
spk = 448
! echo $spk > local/spk
! ./run.sh --stage 1 --stop_stage 10 --skip_train true
```

```python
for set in ['dev', 'sen', 'sp']:
    if os.path.exists(f'text.{set}'):
        os.unlink(f'text.{set}')
    with open(f'data/{set}/wav.scp', 'r') as df:
        lines = df.readlines()
        for line in tqdm(lines):
            id = line.split(' ')[0]
            wav = line.split(' ')[1].rstrip()
            # Check if file exist
            # if not os.path.exists(wav):
            #     print(f"{wav} not exist")
            #     continue            
            # print(f"data:{line}")
            # load audio
            audio = whisper.load_audio(wav)
            audio = whisper.pad_or_trim(audio)
            mel = whisper.log_mel_spectrogram(audio).to(model.device)
            # Get lang probabilities
            _, probs = model.detect_language(mel)
            # print(f"Detected language: {max(probs, key=probs.get)}")
            # make option for model and decode
            options = whisper.DecodingOptions(language='zh', without_timestamps=True, fp16=False)
            result = whisper.decode(model, mel, options)
            text = converter.convert(result.text)
            # print(text)
            with open(f'text.{set}', 'a') as tf:
                tf.write(f"{id} {text}\n")
            !clear
```

```python
tf.close()
```
